# 1: Matplotlib

```python
# Python Visualization Library
# Import the pyplot module from Matplotlib and refer to it as plt.
import matplotlib.pyplot as plt
x = [1,2,3,4,5,6]
y = [10,20,30,40,50,60]
plt.plot(x,y)
plt.show()
plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.plot(x, y, color="green")
plt.plot(x, y, linestyle="-")
plt.plot(x, y, marker="o")
plt.plot(x, y, linewidth=4)
plt.plot(x, y, marker="o", markersize=10)
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[20,30,25,40,50]

plt.figure(figsize=(4,5))

plt.plot(
    x,
    y,
    color="green",
    linestyle="--",
    marker="o",
    linewidth=3,
    markersize=8,
    label="Sales",
    mfc="red",
)

plt.grid()
plt.legend(["sales"])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 2*np.pi, 100)

y = np.cos(x)

plt.plot(x, y, color ="green")
plt.title("cosine wave")
plt.xlabel("x")
plt.ylabel("cos(x)")

plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt 
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [200, 250, 180, 300, 280]
plt.plot(months,sales)
plt.title("marketing")
plt.xlabel("months")
plt.ylabel("sales")
plt.xticks()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [0, 2, 4, 6, 8, 10]
y = [-2, -1, 0, 1, 2, 1]

plt.plot(x, y)

plt.xlim(0, 10)
plt.ylim(-2, 2)

plt.xticks([0, 2, 4, 6, 8, 10])
plt.yticks([-2, -1, 0, 1, 2])

plt.title("Custom Tick Marks")

plt.grid()

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

y1 = x
y2 = [i**2 for i in x]

plt.figure(figsize=(8, 5))

plt.plot(x, y1, label="Linear")
plt.plot(x, y2, label="Square")

plt.title("Linear vs Square")
plt.grid()

plt.legend()

plt.show()
```

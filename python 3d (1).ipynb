# 5. Distance Between 3D Points

```python
# fromula:distance = √((x2-x1)² + (y2-y1)² + (z2-z1)²)
import math #lib
p1 = (0,0,0)
p2 = (3,4,0)
distance = math.sqrt(
    (p2[0]-p1[0])**2 +
    (p2[1]-p1[1])**2 +
    (p2[2]-p1[2])**2
)
print(distance)
     
```

# Distance Formula

```python
import math

x1 = 2
y1 = 3
z1 = 4

x2 = 8
y2 = 7
z2 = 9

distance = math.sqrt(
    (x2 - x1) ** 2 +
    (y2 - y1) ** 2 +
    (z2 - z1) ** 2
)

print(distance)
```

```python
print(round(distance, 2))
```

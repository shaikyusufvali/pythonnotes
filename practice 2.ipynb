```python
n = int(input("Enter the number:"))
Reverse = 0
while n > 0:
    digit = n % 10
    Reverse = Reverse * 10 + digit
    n = n // 10
print(Reverse)
```

```python
import pandas as pd

data = {
    "Name": ["Yusuf", "Rahul", "Priya", "Arjun", "Sneha",
             "Kiran", "Aisha", "Vijay", "Neha", "Ravi"],
    
    "Age": [22, 24, 21, 25, 23, 26, 22, 24, 21, 27],
    
    "Department": ["Python", "Data Science", "Python", "AI", "Data Science",
                   "Python", "AI", "Python", "AI", "Data Science"],
    
    "Salary": [30000, 45000, 32000, 50000, 42000,
               35000, 48000, 38000, 46000, 52000],
    
    "Experience": [1, 2, 1, 3, 2, 2, 1, 2, 1, 4]
}

df = pd.DataFrame(data)

print(df)
```

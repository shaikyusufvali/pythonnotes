```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load Dataset
tips = sns.load_dataset("tips")

# Theme
sns.set_theme(style="whitegrid", palette="deep", font_scale=1)

# Create 4 rows × 2 columns
fig, axes = plt.subplots(4, 2, figsize=(15, 20))

# 1. Scatter Plot
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    ax=axes[0, 0]
)
axes[0, 0].set_title("Scatter Plot")

# 2. Line Plot
sns.lineplot(
    data=tips,
    x="size",
    y="tip",
    marker="o",
    ax=axes[0, 1]
)
axes[0, 1].set_title("Line Plot")

# 3. Bar Plot
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    ax=axes[1, 0]
)
axes[1, 0].set_title("Bar Plot")

# 4. Count Plot
sns.countplot(
    data=tips,
    x="day",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Count Plot")

# 5. Histogram
sns.histplot(
    data=tips,
    x="total_bill",
    kde=True,
    ax=axes[2, 0]
)
axes[2, 0].set_title("Histogram")

# 6. Box Plot
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    ax=axes[2, 1]
)
axes[2, 1].set_title("Box Plot")

# 7. Violin Plot
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    ax=axes[3, 0]
)
axes[3, 0].set_title("Violin Plot")

# 8. Heatmap
corr = tips.corr(numeric_only=True)
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    ax=axes[3, 1]
)
axes[3, 1].set_title("Heatmap")

# Adjust layout
plt.tight_layout()

# Show all plots
plt.show()
```

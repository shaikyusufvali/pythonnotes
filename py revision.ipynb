# . Exception Handling (Must Practice)


```python
while True:

    try:

        a=int(input("Enter First Number: "))
        b=int(input("Enter Second Number: "))

        result=a/b

    except ZeroDivisionError:
        print("Cannot divide by zero")

    except ValueError:
        print("Enter only integers")

    else:
        print("Answer =",result)

    finally:
        print("Execution Finished")

    choice=input("Continue(y/n): ")

    if choice.lower()=="n":
        break
```

```python
class Employee:

    def __init__(self,name):
        self.name=name

    def work(self):
        print(self.name,"is Working")


class Developer(Employee):

    def __init__(self,name,language):
        super().__init__(name)
        self.language=language

    def coding(self):
        print(self.name,"codes in",self.language)


class Tester(Employee):

    def testing(self):
        print(self.name,"is Testing")


d=Developer("Yusuf","Python")
d.work()
d.coding()

t=Tester("Rahul")
t.work()
t.testing()
```

```python
class Father:

    def skills(self):
        print("Programming")


class Mother:

    def hobby(self):
        print("Painting")


class Child(Father,Mother):

    def display(self):
        print("Child Details")


c=Child()

c.display()
c.skills()
c.hobby()
```

```python
class Student:

    college="KHIT"

    def __init__(self,name,age):
        self.name=name
        self.age=age

    def display(self):
        print(self.name,self.age,self.college)

s1=Student("Yusuf",21)
s2=Student("Ali",22)

s1.display()
s2.display()
```

```python
import pandas as pd

data={

    "Name":["A","B","C","D","E"],

    "Age":[20,25,30,22,24],

    "Salary":[30000,45000,60000,35000,50000]

}

df=pd.DataFrame(data)

print(df)

print(df.head())

print(df.tail())

print(df.info())

print(df.describe())

print(df["Salary"].mean())

print(df.sort_values("Salary"))

print(df[df["Salary"]>40000])

df["Bonus"]=df["Salary"]*0.10

print(df)
```

```python
import pandas as pd
import numpy as np

data={

    "Age":[20,np.nan,22,23,np.nan],

    "Salary":[30000,40000,np.nan,50000,45000]

}

df=pd.DataFrame(data)

print(df.isnull().sum())

df["Age"]=df["Age"].fillna(df["Age"].mean())

df["Salary"]=df["Salary"].fillna(df["Salary"].median())

print(df)
```

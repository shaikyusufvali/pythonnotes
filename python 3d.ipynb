# 5. Distance Between 3D Points

```python
# fromula:distance = √((x2-x1)² + (y2-y1)² + (z2-z1)²)
import math #lib
p1 = (0,0,0)
p2 = (3,4,0)
distance = math.sqrt(
    (p2[0]-p1[0])**2 +
    (p2[1]-p1[1])**2 +
    (p2[2]-p1[2])**2
)
print(distance)
     
```
